# Skill-test launcher

Submits `skill-tests/workloads/` on the live cluster. Spark UI: http://localhost:4040

Read `TASK.md` in this folder before changing the jobs.

```python
import runpy
from pathlib import Path

candidates = [
    Path("/skill-tests/launch.py"),
    Path(__file__).resolve().parent / "launch.py" if "__file__" in dir() else None,
    Path.cwd() / "launch.py",
    Path.cwd().parent / "skill-tests" / "launch.py",
]
script = next(p for p in candidates if p is not None and p.exists())
print("running", script)
runpy.run_path(str(script), run_name="__main__")
```
